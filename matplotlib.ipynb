```python
import matplotlib.pyplot as plt
x=[1,2,3,4,5]
y=[2,4,8,8,10]
x1=[9,8,6,5,4]
y1=[5,5,6,8,9]
plt.xlabel("X-axis")
plt.ylabel("y-axis")
plt.xlim(1,10)
plt.ylim(5,15)
plt.title("line plot example")
plt.plot (x,y,label="line-1",color='green',linestyle='dashdot',linewidth=3,marker='D',markerfacecolor='blue',markersize=10)
plt.plot(x1,y1,label="line-2")
plt.legend()
plt.savefig("sample.png")
plt.show()
```
